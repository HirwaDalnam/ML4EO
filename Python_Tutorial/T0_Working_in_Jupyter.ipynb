## Part 3 · The order of cells matters

A notebook **remembers** what you ran. A value stored in one cell can be used in later cells,
but only if the first cell has been run.

### 3.1 Store a value  ▶ RUN
`nir = 0.30` stores the number 0.30 under the name `nir` (near-infrared reflectance). Nothing is shown.

````python
nir = 0.30      # near-infrared reflectance of a pixel
````

### 3.2 Store a second value  ▶ RUN

````python
red = 0.10      # red reflectance of the same pixel
````

### 3.3 Use both values  ▶ RUN
This cell uses `nir` and `red` from the two cells above to compute the NDVI.

````python
ndvi = (nir - red) / (nir + red)
ndvi
````

### 3.4 What happens if you change a cell but do not re-run the others?  ✎ CHANGE
1. Go back to cell **3.2** and change `0.10` to `0.20`.
2. Run **only** cell 3.2.
3. Look at cell 3.3: its old result (0.5) is still shown, because it has not been run again.
4. Now run cell 3.3. The NDVI changes to about 0.2.

**Lesson:** an output always shows the result **at the moment the cell was run**. After a change, re-run the cells below it.

### 3.5 Using a name that does not exist  ▶ RUN
The next cell makes an error **on purpose**: there is no value called `swir` yet.

````python
# This cell makes an error on purpose
swir + nir
````

### 3.6 How to read an error message
Error messages look frightening, but you only need the **last line**:

`NameError: name 'swir' is not defined`

* `NameError` is the **type** of error: Python does not know this name.
* The text after it tells you **which** name.

Typical causes of a `NameError` in the practicals:
* the cell that creates the name has **not been run** (very common after reopening a notebook);
* a **typing mistake**: `ndvi` and `NDVI` are different names for Python.

### 3.7 Restarting cleanly  ✍ EXERCISE
When things get confusing, start again from a clean state. Try this now:
1. In the menu choose **Kernel → Restart Kernel…** and confirm. The notebook forgets all values (the numbers on the left disappear).
2. Click on cell **3.3** and run only that cell. You get a `NameError`: `nir` no longer exists.
3. Choose **Run → Run All Above Selected Cell**, then run cell 3.3 again. It works.

The most useful menu item of all: **Kernel → Restart Kernel and Run All Cells…** It runs every cell from top to bottom, in the right order.
(Cell 3.5 will show its error again: that is expected.)

### 3.8 A cell that takes time  ▶ RUN

## Part 6 · File paths on Windows

Windows writes paths with **backslashes**: `C:\Users\username\Documents\ml4eo\data`.
In Python, a backslash inside text has a special meaning, which causes the most common path errors on Windows.

### 6.1 The backslash problem  ▶ RUN
In Python text, `\n` means "new line" and `\t` means "tab". Look what happens to this path:

````python
print("C:\new_data\table.csv")
````

The path is broken in two: `\n` became a line break and `\t` a tab. Python did not complain; the path is simply wrong, and you get a `FileNotFoundError` later.

### 6.2 An even worse case  ▶ RUN
Every Windows user folder starts with `C:\Users`. In Python, `\U` starts a special code, so this cell fails **on purpose** before it even runs:

````python
# This cell makes an error on purpose
folder = "C:\Users\username\Documents\ml4eo\data"
````

`SyntaxError: (unicode error) 'unicodeescape' codec can't decode bytes …` means: there is a backslash problem in a text.

### 6.3 Three safe ways to write a Windows path  ▶ RUN
All three lines below describe the same folder correctly.

````python
way_1 = "C:/Users/username/Documents/ml4eo/data"        # 1. forward slashes: Windows accepts them
way_2 = r"C:\Users\username\Documents\ml4eo\data"       # 2. an r before the quotes: "raw" text, backslashes are kept
way_3 = "C:\\Users\\username\\Documents\\ml4eo\\data"   # 3. every backslash doubled

print(way_1)
print(way_2)
print(way_3)
````

**Recommendation:** paste a Windows path inside `r"..."` (way 2), or use forward slashes (way 1).

### 6.4 `Path` understands Windows paths  ▶ RUN
`PureWindowsPath` shows how Python handles a Windows path, on any computer. On your Windows PC, `Path(...)` behaves the same.

````python
from pathlib import PureWindowsPath

folder = PureWindowsPath(r"C:\Users\username\Documents\ml4eo\data")
full = folder / "kigali_training_table.csv"

print(full)     # printed as Windows writes it, with backslashes
full            # shown in Python's own notation, with forward slashes
````

### 6.6 Other Windows pitfalls
| Problem | What you see | Solution |
|---|---|---|
| **Hidden file extensions.** Explorer hides `.csv` or `.tiff`, so a file renamed to `table.csv` is really `table.csv.csv` | `FileNotFoundError` although the file "is there" | In Explorer: **View → Show → File name extensions**. Python always needs the full name with extension. |
| **OneDrive "files on demand".** Files with a cloud icon are not really on your computer | Very slow reading, or `OSError`, `PermissionError` | Right-click the `ml4eo` folder → **Always keep on this device**, or keep the course outside OneDrive. |
| **Very long paths** (more than 260 characters in total). Our satellite file names are already about 70 characters long | `FileNotFoundError` for files that exist | Keep `ml4eo` near the top of the drive, e.g. `C:\Users\YourName\Documents\ml4eo`, not deep inside many folders. |
| **Upper and lower case.** Windows ignores case (`B04` = `b04`); other systems (Linux, Google Colab) do not | Code works on your PC but not elsewhere | Always type names exactly as they are written. |
| **Spaces and special characters** in folder names (`My Course (2026)`) | Usually fine in Python, but problems in the Anaconda Prompt | For folders you create, prefer simple names: `ml4eo`, `data`. In the Anaconda Prompt, put a path with spaces in double quotes. |
| **The notebook was moved** out of its folder | `DATA.exists()` is `False` | Keep notebooks inside their `ml4eo` sub-folder. |
| **The file is open in another program** (e.g. a CSV in Excel) | `PermissionError` when writing | Close the file in Excel. |

### 6.9 When a practical cannot find a file: a checklist
