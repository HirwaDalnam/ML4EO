### 2.4 Rules for names
* Letters, digits and `_` only; no spaces, no `-`: `veg_fraction` ✓, `veg fraction` ✗, `veg-fraction` ✗.
* Do not start with a digit: `band_8` ✓, `8band` ✗.
* Upper and lower case are **different**: `NDVI` and `ndvi` are two different variables.
* Choose names that say what the value is: `red`, `nir`, `test_accuracy` rather than `x`, `a`, `t`.

## Part 6 · Comparisons and `if`

### 6.1 Comparisons  ▶ RUN
A comparison gives `True` or `False` (type `bool`).
`>` greater · `<` smaller · `>=` · `<=` · `==` equal (two equal signs!) · `!=` not equal.

````python
ndvi = 0.65
print(ndvi > 0.5)
print(ndvi == 0.65)
print(ndvi != 0.65)
````

### 6.2 Combining conditions  ▶ RUN
`and`: both must be true. `or`: at least one must be true.

````python
print(ndvi > 0.2 and ndvi < 0.5)     # between 0.2 and 0.5?
print(ndvi < 0 or ndvi > 0.6)        # very low or very high?
````

### 6.3 `if` and `else`  ▶ RUN
`if` runs the indented lines only when the condition is `True`; otherwise the lines under `else` run.
Two rules:
* the `if` and `else` lines end with a **colon** `:`;
* the lines that belong to them are **indented** by 4 spaces (Jupyter does this when you press Enter after the colon).

````python
ndvi = 0.65
if ndvi > 0.5:
    print("dense vegetation")
else:
    print("not dense vegetation")
````

✎ **CHANGE:** set `ndvi = 0.3` and run again.

````python
ndvi = 0.35
if ndvi < 0:
    label = "water"
elif ndvi < 0.5:
    label = "sparse vegetation or bare surface"
else:
    label = "dense vegetation"
label
````

### 7.3 Collecting results in a list  ▶ RUN
Start with an empty list `[]` and `.append()` a result in every round.

````python
nir_values = [0.42, 0.30, 0.05]
red_values = [0.06, 0.10, 0.07]
ndvi_list = []
for i in range(3):                      # i = 0, 1, 2
    ndvi_list.append((nir_values[i] - red_values[i]) / (nir_values[i] + red_values[i]))
ndvi_list
````

## Part 8 · Functions

### 8.1 Why functions?
You have written the NDVI formula several times now. A **function** stores a calculation under a name, so that you can use it again with different values.
The practicals use functions such as `load_band(date, band)`, `report(...)` and `show_map(...)`.

### 8.2 Defining a function  ▶ RUN
* `def` starts the definition, followed by the name and, in brackets, the **parameters** (the inputs).
* The line ends with a colon; the body is indented.
* `return` gives the result back.

Running this cell only **defines** the function; nothing is computed yet.

````python
def ndvi(nir, red):
    # Normalised difference vegetation index from near-infrared and red reflectance
    return (nir - red) / (nir + red)
````

### 8.3 Calling a function  ▶ RUN
Now use it, with different values each time.

````python
print(ndvi(0.42, 0.06))
print(ndvi(0.30, 0.10))
print(ndvi(nir=0.05, red=0.07))     # you may also name the inputs
````

### 8.4 `return` or `print`?  ▶ RUN
`return` **gives a value back**, which you can store and use further. `print` only **shows** it on the screen.

````python
value = ndvi(0.42, 0.06)      # the returned value is stored
round(value * 100)            # and can be used in a new calculation
````

### 8.5 A function with a default value  ▶ RUN
A parameter can have a default value, used when you do not give one. Here the date of the September image is the default.

````python
def sentinel_file(band, date="2026-09-03"):
    # File name of one Sentinel-2 band of one date in ml4eo/data
    return f"{date}-00_00_{date}-23_59_Sentinel-2_L2A_{band}_(Raw).tiff"

print(sentinel_file("B08"))
print(sentinel_file("B08", date="2026-01-21"))
````

### 8.6 Your own function  ✍ EXERCISE
NDVI, MNDWI and NDBI are all **normalised differences**: (a − b) / (a + b).
Write a function `nd(a, b)` that returns the normalised difference, then use it to compute the MNDWI of the water pixel from 2.5 (green 0.05, SWIR 0.02).

````python
def nd(a, b):
    return (a - b) / (a + b)                     # ✍ replace None by the formula

mndwi_water = nd(0.05, 0.02)                  # ✍ replace None by a call of nd(...)

check(nd(3, 1) == 0.5, "the formula is (a - b) / (a + b)")
check(mndwi_water is not None and abs(mndwi_water - 3 / 7) < 1e-9, "call nd(0.05, 0.02): green first, SWIR second")
````

## Part 11 · Common errors

| Error (last line of the message) | Typical cause | Fix |
|---|---|---|
| `NameError: name 'x' is not defined` | cell that creates `x` not run; typing mistake; wrong upper/lower case | run the cells above; check the spelling |
| `TypeError: can only concatenate str …` | mixing text and numbers with `+` | use an f-string: `f"NDVI = {ndvi}"` |
| `KeyError: 'B11'` | key (or table column) does not exist | check the exact spelling of the key or column |
| `IndexError: list index out of range` | position too high (counting starts at 0) | check `len(...)` |
| `IndentationError` | missing or wrong indentation after `if`, `for`, `def` | indent the body by 4 spaces |
| `SyntaxError: invalid syntax` | missing colon, bracket or quote; comma instead of a decimal point | look at the line shown (and the one before it) |
| `ZeroDivisionError` | division by 0, e.g. NDVI of a pixel with NIR = Red = 0 | check the input values |
| `ModuleNotFoundError` | library missing, wrong environment | start Jupyter from the `ml4eo` environment |

## Final challenge  ✍ EXERCISE

Three pixels of the Kigali image are stored in a dictionary: pixel name → (red, near-infrared) reflectance.
Write a loop that, for every pixel,
1. computes the NDVI with your function `ndvi(...)`,
2. classifies it with the four classes of 6.6,
3. adds the class to the list `labels`.

Some lines are given. `for name, (red, nir) in pixels.items():` loops over the pairs of the dictionary and unpacks the reflectances into `red` and `nir`.

````python
pixels = {"lake": (0.05, 0.03), "wetland": (0.04, 0.38), "city centre": (0.17, 0.22)}
labels = []
for name, (red, nir) in pixels.items():
    value = ndvi(nir, red)                     # ✍ compute the NDVI with your function
    if value < 0:
        label = "water"
    elif value < 0.2:
        label = "built-up or bare soil"
    elif value < 0.5:
        label = "sparse vegetation"
    else:
        label = "dense vegetation"
    labels.append(label)
    print(f"{name}: NDVI {value:.2f} → {label}")

check(labels == ["water", "dense vegetation", "built-up or bare soil"], "compute value = ndvi(nir, red), use the if/elif/else of 6.6, then labels.append(label)")
````
